```python
import nlpaug.augmenter.word as naw
import nltk
import pandas as pd
import numpy as np

# Install nlpaug and download nltk data if not already done
# These steps are included as relevant cells were deleted from the notebook history
!pip install nlpaug -q
nltk.download('wordnet', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)

# Initialize the synonym augmenter
aug = naw.SynonymAug(aug_src='wordnet')

def augment_product_name(product_name):
    """Applies synonym augmentation to a single product name."""
    try:
        # nlpaug expects a list, so wrap the string
        augmented_text = aug.augment(product_name)
        return augmented_text[0] # Return the first augmented result
    except Exception as e:
        # If augmentation fails (e.g., no synonyms found), return original
        return product_name

# Select a small number of rows to augment as an example
# Let's augment the first 10 rows for demonstration purposes
sample_to_augment = df.head(10).copy()

# Apply augmentation to the 'Product' column of the sample
augmented_rows_list = []
for index, row in sample_to_augment.iterrows():
    augmented_row = row.copy()
    original_product = augmented_row['Product']
    augmented_product = augment_product_name(original_product)

    # Only add if a different augmented product was generated
    if augmented_product != original_product:
        augmented_row['Product'] = augmented_product
        augmented_rows_list.append(augmented_row)

# Convert the list of augmented rows to a DataFrame
df_augmented_sample = pd.DataFrame(augmented_rows_list)

# Concatenate the original DataFrame with the newly augmented sample rows
# Using `ignore_index=True` ensures the new DataFrame has a continuous index
df_combined = pd.concat([df, df_augmented_sample], ignore_index=True)

print(f"Original DataFrame shape: {df.shape}")
print(f"Number of new augmented sample rows added: {len(df_augmented_sample)}")
print(f"Combined DataFrame shape after augmenting sample rows: {df_combined.shape}")

# Display the head and tail of the combined DataFrame to show augmented data
print("\nHead of the combined DataFrame:")
display(df_combined.head(15))

print("\nTail of the combined DataFrame (showing some augmented rows):")
display(df_combined.tail(15))

# Update the main DataFrame 'df' if you want to keep the augmented data for further steps
df = df_combined
```

```python
print("\n--- Feature Selection using SelectKBest (Mutual Information) ---")
# Select the top k features
k_best_selector = SelectKBest(score_func=mutual_info_classif, k=5) # Select top 5 features
X_kbest = k_best_selector.fit_transform(X, y)

# Get the selected feature names
selected_features_mi = X.columns[k_best_selector.get_support(indices=True)].tolist()
print(f"Top {len(selected_features_mi)} features selected by mutual_info_classif: {selected_features_mi}")

# Display scores of all features
feature_scores_mi = pd.Series(k_best_selector.scores_, index=X.columns)
print("\nMutual Information scores for all features:")
print(feature_scores_mi.sort_values(ascending=False))
```

```python

df['Avg_Price_Per_Unit'] = df.apply(lambda row: row['Total Sales'] / row['Quantity'] if row['Quantity'] != 0 else row['Price'], axis=1)
```

#### 4. Interaction feature: Price and Encoded Category

```python

df['Price_Category_Interaction'] = df['Price'] * df['Category_Encoded']
```

#### 5. Polynomial feature: Price Squared

```python

df['Price_Squared'] = df['Price']**2
```

Training and evaluation


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from google.colab import files
import pandas as pd

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())

df.columns = df.columns.str.strip()

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())

# STEP 2: Clean the dataset
df = df.dropna(subset=["Status"]).copy()
df["Status"] = df["Status"].astype(str).str.strip()

# STEP 3: Process the date column
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(
        df["Date"],
        format="%d-%m-%y",
        errors="coerce"
    )

    df["Order_Month"] = df["Date"].dt.month
    df["Order_Day"] = df["Date"].dt.day
    df["Order_DayOfWeek"] = df["Date"].dt.dayofweek
    df["Is_Weekend"] = (
        df["Order_DayOfWeek"] >= 5
    ).astype(float)

    df = df.drop(columns=["Date"])

# STEP 4: Remove identifiers that should not be predictors
df = df.drop(
    columns=["Order ID", "Customer Name"],
    errors="ignore"
)

# Remove rows with missing target labels
X = df.drop(columns=["Status"])
y = df["Status"]

# Remove rows where no usable feature is available
valid_rows = X.notna().any(axis=1)
X = X.loc[valid_rows].copy()
y = y.loc[valid_rows].copy()

# Remove constant columns
constant_cols = [
    col for col in X.columns if X[col].nunique(dropna=True) <= 1
]
X = X.drop(columns=constant_cols)

# STEP 5: Identify numerical and categorical columns
numeric_cols = X.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_cols = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumerical features:", numeric_cols)
print("Categorical features:", categorical_cols)
print("\nOrder status distribution:")
print(y.value_counts())

if y.nunique() < 2:
    raise ValueError("Training requires at least two target classes.")

# STEP 6: Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

# STEP 7: Preprocess numerical and categorical features
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

transformers = []

if numeric_cols:
    transformers.append(
        ("numeric", numeric_pipeline, numeric_cols)
    )

if categorical_cols:
    transformers.append(
        ("categorical", categorical_pipeline, categorical_cols)
    )

preprocessor = ColumnTransformer(transformers=transformers)

# STEP 8: Build the Random Forest model
model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    ))
])

# STEP 9: Train the model
model.fit(X_train, y_train)

print("\nModel training completed successfully!")

# STEP 10: Make predictions
y_pred = model.predict(X_test)

# STEP 11: Evaluate model performance
accuracy = accuracy_score(y_test, y_pred)

print("\n========== MODEL EVALUATION ==========")
print(f"Accuracy: {accuracy:.4f}")
print(f"Accuracy: {accuracy * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))

# STEP 12: Plot the confusion matrix
labels = model.named_steps["classifier"].classes_

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=labels
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Random Forest - Amazon Order Status")
plt.tight_layout()
plt.show()

# STEP 13: Show a sample prediction
sample = X_test.iloc[[0]]
actual = y_test.iloc[0]
predicted = model.predict(sample)[0]

print("\n========== SAMPLE PREDICTION ==========")
print("Actual Status:", actual)
print("Predicted Status:", predicted)

# STEP 14: Show feature importance
feature_names = model.named_steps[
    "preprocessor"
].get_feature_names_out()

importances = model.named_steps[
    "classifier"
].feature_importances_

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
}).sort_values("Importance", ascending=False)

print("\nTop 10 Important Features:")
display(importance_df.head(10))

top_features = importance_df.head(10).sort_values("Importance")

plt.figure(figsize=(9, 5))
plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)
plt.xlabel("Feature Importance")
plt.ylabel("Feature")
plt.title("Top 10 Features - Random Forest")
plt.tight_layout()
plt.show()

print("\nAll training and evaluation steps completed!")
```
